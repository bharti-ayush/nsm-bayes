## gnk

```python
m = 100_000
```

```python
# posterior sampling
num_sample_per_param = 15
num_posterior = 500
warmup = 500
num_posterior_samples = num_posterior + warmup
```

```python
budget_posterior_sampling = num_posterior_samples * num_sample_per_param
```

```python
# picking beta
n_steps_theta_hat = 200
bootstrap_size = 20
grid_size_beta = 3
num_posterior_bootstrap = 200
num_sample_per_param_bootstrap = 7
```

```python
budget_theta_hat = n_steps_theta_hat * num_sample_per_param_bootstrap
```

```python
budget_picking_beta = budget_theta_hat + grid_size_beta * bootstrap_size * num_posterior_bootstrap * num_sample_per_param_bootstrap
print("Budget used for picking beta: ", budget_picking_beta)
```

```python
print("budget constraint: ", m)
print("num_sampl_per_param:", num_sample_per_param)
print("grid_size_beta:", grid_size_beta)
print("bootstrap_size:", bootstrap_size)
print("num_posterior_bootstrap:", num_posterior_bootstrap)
print("num_sample_per_param_bootstrap:", num_sample_per_param_bootstrap)
print("--------------------------------")
print("budget posterior sampling:", budget_posterior_sampling)
print("budget picking beta:", budget_picking_beta)
print("budget: ", budget_posterior_sampling + budget_picking_beta)
```
